```python
# -*- coding: utf-8 -*-
"""637 nm RAW near-field MFD and spatial-aperture power -- ONE CELL.

USAGE
1. Enter RAW_INPUT (one RAW file, a folder, or "" for the current folder)
   and SENSOR_POWER_UW (total measured optical power in microwatts).
2. Confirm RAW format, camera pixel size, magnification/object-plane pixel scale,
   and whether the power meter was before or after the ND filter.
3. Run this cell. A file such as 50C_W1.raw creates a folder called 50C_W1.
   batch_summary.csv in the input folder combines all results and failures.
   In batch mode, use POWER_BY_FILENAME_UW when meter readings differ by file.

WHAT IS MEASURED
- The Gaussian intensity is I=A*exp(-2*d.T*inv(S)*d), plus a background plane.
  Each modeled camera pixel integrates the intensity over its area. Fitting
  uses the unsmoothed RAW counts. Filtering is used ONLY for initial detection.
- mfd_x_um and mfd_y_um are the full 1/e^2 intensity widths of horizontal and
  vertical cuts through the fitted center. This threshold is 13.5335% of the
  background-subtracted peak INTENSITY, not 13.5335% of field amplitude.
- Major/minor axes and projected widths have different names and are saved
  separately. For a rotated ellipse, they need not equal the X/Y center cuts.
- Direct 1/e^2 crossings, finite-aperture D4sigma, residuals, and ROI/robust-fit
  sensitivity are checks. They do not establish that a field is single mode.
  The D4sigma calculation here uses a local aperture; it is not an assertion
  of ISO 11146 compliance or a Petermann-II fiber MFD measurement.

POWER AND INSERTION LOSS
- Signed background-corrected counts are summed. Negative noise is NOT set
  to zero: doing so over a large camera field introduces positive bias.
- The default power ellipse has radii 2 times the fitted 1/e^2 radii. It
  contains 1-exp(-8)=99.9665% of an ideal 2-D Gaussian. The 1/e^2 ellipse alone
  contains only 1-exp(-2)=86.4665%. No Gaussian tail correction is forced on
  real data. The aperture sweep shows sensitivity to the selected region.
- Core-region power at the meter plane = measured total power * signed
  aperture counts / signed total-view counts. The meter and camera must
  collect the SAME optical field, including any halo/stray light in the total.
- With uniform ND attenuation, the spatial fraction is unchanged. If the
  meter reading is BEFORE the ND, do not multiply the result by 1000.
  If it is AFTER the ND and you want BEFORE-ND power, enter the independently
  measured filter transmission at 637 nm. Nominal ND3 is not calibration.
- This is power in a spatial region, not a modal decomposition. Overlapping
  guided modes and stray light cannot be uniquely separated by one intensity
  image. The infinite-Gaussian component is saved only as a model diagnostic.
- Optional apparent IL=-10*log10(core power before ND / reference input power).
  Define matching reference planes. Collection/relay losses remain included
  unless independently accounted for. This alone is not propagation loss.

ACCURACY / CALIBRATION
Use a linear RAW export, fixed exposure/gain, and a conjugate image of the
waveguide output facet. JPG is not accepted for numerical measurement.
Provide beam-blocked dark RAWs with identical acquisition settings whenever
possible. A dark frame preserves electronic structure; an edge-only estimate
cannot distinguish broad optical light from an electronic baseline.
Check that your camera has a calibrated, linear response at 637 nm. Verify
3.45 um pixels and 6.9x magnification for your actual setup; they are presets,
not measurements made by this notebook. Defocus, collection NA, imaging PSF,
pixel scale, black-level clipping, and meter/collection mismatch limit absolute
accuracy. A measured Gaussian optical PSF covariance can optionally be
subtracted; no PSF is guessed. Pixel integration is already in the fit.

Fit u1sigma values describe conditional fitting precision. Even the optional
fit-and-scale uncertainty omits PSF, calibration drift, model mismatch and
other systematics. The power noise indicator assumes independent pixel noise
and omits spatial correlations, aperture-fit and meter calibration uncertainty.
Review flags and repeat acquisitions; no guaranteed percent accuracy follows
from a good fit. Review thresholds are practical diagnostics, not standards.

OUTPUTS
Per file: summary CSV, measured/fitted profiles CSV, aperture sweep CSV,
ROI/robust-fit sensitivity CSV, PNG/PDF diagnostics, NPZ fit/crop arrays, and
JSON metadata. Saved MFD estimates can remain numeric when flagged; check
mfd_quality before using them. Invalid power values are suppressed. RAW input
files are read only. Rerunning replaces generated outputs of the same names.
Dependencies: numpy, scipy, pandas, matplotlib; no other packages required.

637 NM FILE READER
The supplied DeviceData_FV_#001.raw has a 54-byte BMP-style header and
2448 x 2048 unsigned 16-bit grayscale samples. Its negative stored height
means top-down rows. This is a camera-specific RAW container, not a color
RGB555 bitmap; ordinary BMP display decoders must not be used for counts.
Auto mode recognizes the observed custom-header structure and reads signed
height, dimensions, pixel offset and row padding. It falls back to explicit
headerless settings for other RAW files. No undocumented header field is
used for baseline correction, pixel pitch or exposure.
The 3.45 um pixel pitch and 12-bit ceiling are BC207-family PRESETS supported
by the matching resolution and manufacturer specifications; verify your model,
binning and magnification before treating the micrometer values as calibrated.
Leave SENSOR_POWER_UW=None for MFD-only analysis. Never reuse a power value
from a different image or wavelength without measuring it.

METHOD REFERENCES (primary sources)
https://www.ophiropt.com/n/beam-width-measurement-accuracy
https://www.corning.com/microsites/coc/oem/documents/specialty-fiber/WP7071-Mode-Field-Diam-and-Eff-Area.pdf
https://www.thorlabs.us/NewGroupPage9.cfm?ObjectGroup_ID=3483
https://learn.microsoft.com/en-us/windows/win32/api/wingdi/ns-wingdi-bitmapinfoheader
"""
from pathlib import Path
import hashlib
import struct
import json
import platform
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
from scipy.ndimage import gaussian_filter, median_filter, label, map_coordinates
from scipy.optimize import least_squares

# =============================================================================
# USER INPUTS -- usually change only these first two values
# =============================================================================
RAW_INPUT = r"DeviceData_FV_#001.raw"  # one RAW file OR folder; "" = current folder
SENSOR_POWER_UW = None             # measured µW; None = MFD/fraction only, no absolute power
POWER_BY_FILENAME_UW = {}           # optional batch overrides: {"50C_W1.raw": 77.0}

WAVELENGTH_NM = 637.0
RAW_STORAGE_FORMAT = "auto"        # "auto", "headerless", or "bmp_gray16"
# Fallback settings below apply ONLY to headerless files.
RAW_WIDTH, RAW_HEIGHT = 2448, 2048
RAW_DTYPE = "uint16"               # unpacked integer data, not packed Bayer/color
RAW_VALID_BITS = 12                # BC207 preset: 12-bit camera values in 16-bit words
RAW_BYTE_ORDER = "native"          # "native", "little", or "big"
RAW_HEADER_BYTES = 0               # exact header size, if any; packed RAW unsupported
SATURATION_COUNT = None            # None -> 2^min(RAW_VALID_BITS, storage_bits)-1

CAMERA_PIXEL_SIZE_UM = 3.45         # BC207 preset; VERIFY camera model/binning!
MAGNIFICATION = 6.9                # calibrated image magnification at 637 nm
# Prefer measured object-plane scale when available; overrides pixel size / M.
SAMPLE_UM_PER_PIXEL_XY = None       # e.g. (0.5000, 0.5000), NOT sensor-plane scale
SCALE_RELATIVE_U_1SIGMA = None      # e.g. 0.01 for 1%; None = unknown, not zero
# Optional optical-blur correction, only for a separately CALIBRATED Gaussian PSF.
# This is the PSF intensity covariance referred to the object plane, in µm².
# For an aligned PSF with 1/e² diameters dx,dy: diag((dx/4)**2, (dy/4)**2).
PSF_COVARIANCE_OBJECT_UM2 = None

# Camera data must be linear, with gamma/auto contrast/auto exposure disabled.
# Capture beam-blocked RAWs at the SAME exposure, gain, ND, ROI and temperature.
DARK_RAW_PATHS = []                # [r"dark1.raw", r"dark2.raw", ...]; averaged
BAD_PIXEL_MASK_NPY = None          # optional calibrated boolean mask: True = bad
FLAT_FIELD_NPY = None              # optional normalized RELATIVE detector response
# Flat field should be dark-corrected, positive, and measured in the same setup.
REMOVE_EDGE_BACKGROUND_PLANE = None  # None: fit/subtract edge plane ONLY without darks
# True forces an additional edge-plane correction; False uses dark subtraction only.
# Do not remove an edge plane that actually represents light collected by the meter.
BACKGROUND_EDGE_FRACTION = 0.12

# By default report power at the plane where the meter reading was taken.
POWER_METER_PLANE = "before_nd"    # "before_nd" or "after_nd"; set to your setup
ND_TRANSMISSION_AT_637 = None      # measured fraction, e.g. 0.0012; never inferred
# OD=3 nominally means T=0.001, but its actual value at 637 nm must be measured.
REFERENCE_INPUT_POWER_BEFORE_ND_UW = None  # optional: input power at chosen link plane
# IL includes collection/relay optics unless independently corrected/calibrated.

# Power meter and camera must collect the SAME optical field/aperture.
POWER_ROI_XYXY = None              # None = full frame; otherwise (x1,y1,x2,y2)
BEAM_CENTER_XY = None              # None = brightest spot; otherwise (x,y) in pixels
FIT_INITIAL_HALF_SIZE_PX = 65
FIT_MIN_HALF_SIZE_PX = 18
FIT_MAX_HALF_SIZE_PX = 160
POWER_ELLIPSE_SCALE = 2.0           # 2 x fitted 1/e^2 radii; includes Gaussian tails
APERTURE_SWEEP_SCALES = (1.0, 1.5, 2.0, 2.5)
APERTURE_SUBPIXELS = 4             # fractional boundary pixels for power integration
CHECK_FIT_SENSITIVITY = True       # refits at smaller/larger ROI and robust loss
SAVE_OUTPUTS = True
SHOW_PLOTS = True                 # False speeds up batch notebook display
SAVE_PDF = True
PNG_DPI = 300

# =============================================================================
# INPUT / BACKGROUND -- signed counts are retained throughout the analysis
# =============================================================================
def load_raw_info(path):
    """Return linear grayscale counts and the observed storage metadata."""
    path = Path(path)
    if RAW_STORAGE_FORMAT not in ("auto", "headerless", "bmp_gray16"):
        raise ValueError("RAW_STORAGE_FORMAT must be auto, headerless or bmp_gray16.")
    with path.open("rb") as stream:
        header = stream.read(54)
    has_bmp_header = len(header) >= 54 and header[:2] == b"BM"
    if RAW_STORAGE_FORMAT == "bmp_gray16" or (RAW_STORAGE_FORMAT == "auto" and has_bmp_header):
        if not has_bmp_header:
            raise ValueError("Selected bmp_gray16 but no valid BM header was found.")
        _, declared_size, reserved1, reserved2, offset = struct.unpack_from("<2sIHHI", header, 0)
        dib, width, stored_height, planes, bits, compression = struct.unpack_from("<IiiHHI", header, 14)
        if dib < 40 or planes != 1 or bits != 16 or compression != 0:
            raise ValueError("Only uncompressed 16-bit grayscale camera RAW wrappers are supported. "
                             "Color, compressed and packed data cannot be analyzed as intensity.")
        if RAW_STORAGE_FORMAT == "auto" and not (declared_size == 0 and (reserved1 or reserved2)):
            raise ValueError("BM header may describe RGB555 color, not camera grayscale. "
                             "Use bmp_gray16 only if the camera export is confirmed grayscale.")
        height = abs(stored_height)
        if width <= 0 or height <= 0 or offset < 14+dib:
            raise ValueError("Invalid dimensions or pixel offset in RAW header.")
        stride = ((width*16+31)//32)*4
        expected = offset + stride*height
        if path.stat().st_size != expected:
            raise ValueError(f"{path.name}: wrapper expects {expected} bytes; "
                             f"found {path.stat().st_size}. Unsupported layout/trailer.")
        data = np.fromfile(path, dtype=np.uint8, offset=offset).reshape(height, stride)
        # Preserve the actual grayscale 16-bit samples; discard row padding only.
        image = np.ascontiguousarray(data[:, :width*2]).view("<u2").reshape(height, width)
        if stored_height > 0:
            image = image[::-1, :]
        info = dict(format="camera_bmp_gray16", width_px=width, height_px=height,
                    storage_bits=16, dtype="uint16", byte_order="little", header_bytes=offset,
                    stride_bytes=stride, source_row_order="top_down" if stored_height<0 else "bottom_up",
                    header_reserved1_raw=reserved1, header_reserved2_raw=reserved2)
        return image.astype(np.float64), info
    dt = np.dtype(RAW_DTYPE)
    if dt.kind != "u" or dt.itemsize not in (1, 2):
        raise ValueError("Use unpacked uint8 or uint16 RAW data.")
    if RAW_BYTE_ORDER not in ("native", "little", "big"):
        raise ValueError("RAW_BYTE_ORDER must be native, little or big.")
    dt = dt.newbyteorder({"native":"=", "little":"<", "big":">"}[RAW_BYTE_ORDER])
    expected = RAW_HEADER_BYTES + RAW_WIDTH*RAW_HEIGHT*dt.itemsize
    if path.stat().st_size != expected:
        raise ValueError(f"{path.name}: {path.stat().st_size} bytes; expected {expected}. "
                         "Check dimensions, header, dtype, or packed format.")
    image = np.fromfile(path, dtype=dt, offset=RAW_HEADER_BYTES).reshape(RAW_HEIGHT, RAW_WIDTH)
    info = dict(format="headerless", width_px=RAW_WIDTH, height_px=RAW_HEIGHT,
                storage_bits=8*dt.itemsize, dtype=dt.name, byte_order=RAW_BYTE_ORDER,
                header_bytes=RAW_HEADER_BYTES, stride_bytes=RAW_WIDTH*dt.itemsize,
                source_row_order="top_down")
    return image.astype(np.float64), info


def load_raw(path):
    return load_raw_info(path)[0]


def robust_sigma(x):
    x = np.asarray(x)
    x = x[np.isfinite(x)]
    if not x.size:
        return 0.0
    return float(1.4826 * np.median(np.abs(x - np.median(x))))


def pixel_scale():
    scale = np.asarray(SAMPLE_UM_PER_PIXEL_XY if SAMPLE_UM_PER_PIXEL_XY is not None
                       else (CAMERA_PIXEL_SIZE_UM / MAGNIFICATION,) * 2, dtype=float)
    if scale.shape != (2,) or not np.all(np.isfinite(scale) & (scale > 0)):
        raise ValueError("Object-plane X/Y pixel scales must be positive.")
    return scale


def roi_bounds(shape, center, half):
    h, w = shape
    x, y = center
    return (max(0, int(np.floor(x - half))), max(0, int(np.floor(y - half))),
            min(w, int(np.ceil(x + half + 1))), min(h, int(np.ceil(y + half + 1))))


def rectangle_mask(shape, bounds=None):
    h, w = shape
    bounds = (0, 0, w, h) if bounds is None else bounds
    if len(bounds) != 4 or any(int(v) != v for v in bounds):
        raise ValueError("ROI must contain four integer pixel bounds.")
    x1, y1, x2, y2 = map(int, bounds)
    if not (0 <= x1 < x2 <= w and 0 <= y1 < y2 <= h):
        raise ValueError("ROI must lie inside the image; x2 and y2 are exclusive.")
    mask = np.zeros(shape, dtype=bool)
    mask[y1:y2, x1:x2] = True
    return mask


def edge_plane(image, good):
    """Robust least-squares plane on edges; symmetric rejection, no zero clipping."""
    h, w = image.shape
    yy, xx = np.indices(image.shape)
    ex, ey = max(2, int(w * BACKGROUND_EDGE_FRACTION)), max(2, int(h * BACKGROUND_EDGE_FRACTION))
    edge = ((xx < ex) | (xx >= w-ex) | (yy < ey) | (yy >= h-ey)) & good
    # Deterministic subsample for a quick, reproducible baseline fit.
    ids = np.flatnonzero(edge)
    if ids.size < 100:
        raise ValueError("Too few valid background-edge pixels.")
    ids = ids[::max(1, int(np.ceil(ids.size / 60000)))]
    xn, yn = (xx.ravel()[ids] - (w-1)/2) / w, (yy.ravel()[ids] - (h-1)/2) / h
    X = np.column_stack((np.ones(len(ids)), xn, yn))
    z = image.ravel()[ids]
    keep = np.abs(z - np.median(z)) <= max(4 * robust_sigma(z), 1.0)
    for _ in range(8):
        if keep.sum() < 100:
            raise ValueError("Background fit rejected too many pixels.")
        coeff = np.linalg.lstsq(X[keep], z[keep], rcond=None)[0]
        residual = z - X @ coeff
        sig = max(robust_sigma(residual[keep]), 0.25)
        new_keep = np.abs(residual - np.median(residual[keep])) < 4.0 * sig
        if np.array_equal(keep, new_keep):
            break
        keep = new_keep
    coeff = np.linalg.lstsq(X[keep], z[keep], rcond=None)[0]
    residual = z - X @ coeff
    # Conditional covariance; spatially correlated electronics/stray light are NOT included.
    variance = float(np.mean(residual[keep]**2))
    covariance = variance * np.linalg.pinv(X[keep].T @ X[keep])
    plane = coeff[0] + coeff[1] * (xx-(w-1)/2)/w + coeff[2] * (yy-(h-1)/2)/h
    return plane, coeff, covariance, max(robust_sigma(residual[keep]), 0.25), edge


def correct_frame(raw, dark=None, flat=None, bad=None):
    good = np.isfinite(raw) if bad is None else np.isfinite(raw) & ~bad
    z = raw.copy() if dark is None else raw - dark
    if flat is not None:
        good &= np.isfinite(flat) & (flat > 0)
        z = np.divide(z, flat, out=np.full_like(z, np.nan), where=good)
    plane, coeff, cov, noise, edge = edge_plane(z, good)
    remove_plane = dark is None if REMOVE_EDGE_BACKGROUND_PLANE is None else bool(REMOVE_EDGE_BACKGROUND_PLANE)
    if remove_plane:
        corrected = z - plane
    else:
        if dark is None:
            raise ValueError("Without dark frames, keep edge background correction enabled.")
        corrected = z
        coeff, cov = np.zeros(3), np.zeros((3, 3))
    # Fill only documented bad pixels for power integration and visualization.
    # Fits exclude them. Replacements are approximate and recorded in the flags.
    if not good.all():
        fill = median_filter(np.nan_to_num(corrected, nan=0.0), size=3)
        corrected[~good] = fill[~good]
    return corrected, good, coeff, cov, noise, edge


# =============================================================================
# PIXEL-AREA GAUSSIAN FIT -- 2-D ellipse + local residual background plane
# =============================================================================
# Three-point Gauss-Legendre quadrature in each pixel dimension.
_quad_x, _quad_w = np.polynomial.legendre.leggauss(3)
_quad_x, _quad_w = _quad_x / 2, _quad_w / 2


def shape_matrix(p):
    wx, wy = np.exp(p[3:5])
    rho = np.tanh(p[5])
    return np.array([[wx**2, rho * wx * wy], [rho * wx * wy, wy**2]])


def gaussian_point(p, x, y):
    wx, wy = np.exp(p[3:5])
    rho = np.tanh(p[5])
    dx, dy = (x-p[1])/wx, (y-p[2])/wy
    q = (dx*dx + dy*dy - 2*rho*dx*dy) / (1-rho*rho)
    return p[0] * np.exp(-2*q)


def gaussian_pixel(p, x, y):
    out = np.zeros_like(x, dtype=float)
    for dx, a in zip(_quad_x, _quad_w):
        for dy, b in zip(_quad_x, _quad_w):
            out += a * b * gaussian_point(p, x+dx, y+dy)
    return out


def background_local(p, x, y, anchor, half):
    return p[6] + p[7]*(x-anchor[0])/half + p[8]*(y-anchor[1])/half


def metrics(p):
    S = shape_matrix(p)
    Q = np.linalg.inv(S)
    scale = pixel_scale()
    Ss = np.diag(scale) @ S @ np.diag(scale)
    ev, U = np.linalg.eigh(Ss)
    v = U[:, 1]
    angle = (np.degrees(np.arctan2(v[1], v[0])) + 90) % 180 - 90
    return {
        # Camera-axis cuts THROUGH THE FITTED CENTER, not projected widths.
        "mfd_x_um": float(2/np.sqrt(Q[0, 0]) * scale[0]),
        "mfd_y_um": float(2/np.sqrt(Q[1, 1]) * scale[1]),
        "mfd_minor_um": float(2*np.sqrt(ev[0])),
        "mfd_major_um": float(2*np.sqrt(ev[1])),
        "mfd_projected_x_um": float(2*np.sqrt(Ss[0, 0])),
        "mfd_projected_y_um": float(2*np.sqrt(Ss[1, 1])),
        "major_axis_angle_deg": float(angle),
        "ellipticity_major_over_minor": float(np.sqrt(ev[1]/ev[0])),
    }


def initial_peak(image, good):
    if BEAM_CENTER_XY is not None:
        x, y = map(float, BEAM_CENTER_XY)
        if not (0 <= x < image.shape[1] and 0 <= y < image.shape[0]):
            raise ValueError("BEAM_CENTER_XY is outside the frame.")
        return x, y
    # Median/smoothing are used ONLY to locate the beam, never to fit its width.
    detect = gaussian_filter(median_filter(image, size=3), sigma=1.0)
    detect[~good] = -np.inf
    y, x = np.unravel_index(np.argmax(detect), image.shape)
    return float(x), float(y)


def fit_gaussian(image, good, center, half, noise, seed=None, robust=False):
    bounds = roi_bounds(image.shape, center, half)
    x1, y1, x2, y2 = bounds
    crop = image[y1:y2, x1:x2]
    valid = good[y1:y2, x1:x2]
    yy, xx = np.mgrid[y1:y2, x1:x2]
    x, y, z = xx[valid].astype(float), yy[valid].astype(float), crop[valid]
    if len(z) < 100:
        raise ValueError("Too few fit pixels.")
    base = float(np.median(z))
    peak = float(np.max(median_filter(crop, size=3)) - base)
    if peak < 5 * noise:
        raise ValueError("Beam signal is too weak for a stable Gaussian fit.")
    if seed is None:
        smooth = gaussian_filter(median_filter(crop, size=3), 1.0)
        selected, _ = label(smooth > base + max(0.15*peak, 3*noise))
        near_x = np.clip(int(round(center[0]-x1)), 0, crop.shape[1]-1)
        near_y = np.clip(int(round(center[1]-y1)), 0, crop.shape[0]-1)
        region_id = selected[near_y, near_x]
        weights = np.maximum(smooth-base, 0) * (selected == region_id) if region_id else np.maximum(smooth-base, 0)
        total = weights.sum()
        if total <= 0:
            raise ValueError("No isolated positive beam was detected.")
        cx, cy = np.sum(weights*xx)/total, np.sum(weights*yy)/total
        wx = np.clip(2*np.sqrt(np.sum(weights*(xx-cx)**2)/total), 2.0, half/1.2)
        wy = np.clip(2*np.sqrt(np.sum(weights*(yy-cy)**2)/total), 2.0, half/1.2)
        rho = np.clip(4*np.sum(weights*(xx-cx)*(yy-cy))/total / (wx*wy), -0.8, 0.8)
        guess = np.array([peak, cx, cy, np.log(wx), np.log(wy), np.arctanh(rho), base, 0., 0.])
    else:
        guess = seed.copy()
    # Center constrained to the selected crop; positive-definite ellipse by construction.
    lower = [0, x1, y1, np.log(0.6), np.log(0.6), -2.5, -4*peak, -4*peak, -4*peak]
    upper = [8*max(peak, noise), x2-1, y2-1, np.log(2*half), np.log(2*half), 2.5, 4*peak, 4*peak, 4*peak]
    guess = np.clip(guess, np.array(lower)+1e-8, np.array(upper)-1e-8)

    def fun(p):
        return (gaussian_pixel(p, x, y) + background_local(p, x, y, center, half) - z) / noise

    opt = least_squares(fun, guess, bounds=(lower, upper), x_scale="jac",
                        loss="soft_l1" if robust else "linear", f_scale=2.0,
                        ftol=1e-8, xtol=1e-8, gtol=1e-8, max_nfev=300)
    if not opt.success:
        raise ValueError("Gaussian fit did not converge: " + opt.message)
    p = opt.x
    residual = fun(p) * noise
    # Conditional 1-sigma fit precision under independent, equal-variance errors.
    # This is not total metrology uncertainty; see ROI checks and scale uncertainty.
    dof = max(1, len(z)-len(p))
    covariance = np.linalg.pinv(opt.jac.T @ opt.jac) * float(np.sum(opt.fun**2)/dof)
    core = gaussian_point(p, x, y) > p[0]*np.exp(-2)
    core_rms = float(np.sqrt(np.mean(residual[core]**2))) if core.any() else np.inf
    return dict(p=p, covariance=covariance, bounds=bounds, anchor=center, half=half,
                nrmse_core=core_rms/max(p[0], 1e-12), rms_counts=float(np.sqrt(np.mean(residual**2))),
                at_bound=bool(np.any(opt.active_mask)), robust=robust)


def adaptive_fit(image, good, center, noise):
    fit = fit_gaussian(image, good, center, FIT_INITIAL_HALF_SIZE_PX, noise)
    for _ in range(3):
        p = fit["p"]
        half = int(np.clip(np.ceil(3.0*np.sqrt(np.max(np.linalg.eigvalsh(shape_matrix(p))))),
                           FIT_MIN_HALF_SIZE_PX, FIT_MAX_HALF_SIZE_PX))
        new_center = (float(p[1]), float(p[2]))
        if abs(half-fit["half"]) < 2 and np.linalg.norm(np.array(new_center)-fit["anchor"]) < 1:
            break
        fit = fit_gaussian(image, good, new_center, half, noise, seed=p)
    return fit


def fit_precision(fit):
    p, C = fit["p"], fit["covariance"]
    result = {}
    for name in ("mfd_x_um", "mfd_y_um", "mfd_minor_um", "mfd_major_um"):
        g = np.zeros(len(p))
        for j in range(3, 6):
            dp = np.zeros(len(p)); dp[j] = 1e-5
            g[j] = (metrics(p+dp)[name]-metrics(p-dp)[name])/(2e-5)
        u = float(np.sqrt(max(0.0, g @ C @ g)))
        result[name.replace("_um", "_fit_u1sigma_um")] = u
        if SCALE_RELATIVE_U_1SIGMA is not None:
            result[name.replace("_um", "_fit_and_scale_u1sigma_um")] = float(
                np.hypot(u, metrics(p)[name]*SCALE_RELATIVE_U_1SIGMA))
    return result


def calibrated_psf_correction(p):
    if PSF_COVARIANCE_OBJECT_UM2 is None:
        return {}
    psf = np.asarray(PSF_COVARIANCE_OBJECT_UM2, dtype=float)
    if psf.shape != (2, 2) or not np.allclose(psf, psf.T) or np.min(np.linalg.eigvalsh(psf)) < 0:
        raise ValueError("Calibrated PSF covariance must be a symmetric positive-semidefinite 2x2 matrix.")
    D = np.diag(pixel_scale())
    intrinsic_S = D @ shape_matrix(p) @ D - 4*psf
    if np.min(np.linalg.eigvalsh(intrinsic_S)) <= 0:
        raise ValueError("PSF correction is unresolved: inferred intrinsic covariance is not positive definite.")
    Q = np.linalg.inv(intrinsic_S)
    ev = np.linalg.eigvalsh(intrinsic_S)
    return dict(psf_corrected_mfd_x_um=float(2/np.sqrt(Q[0, 0])),
                psf_corrected_mfd_y_um=float(2/np.sqrt(Q[1, 1])),
                psf_corrected_mfd_minor_um=float(2*np.sqrt(ev[0])),
                psf_corrected_mfd_major_um=float(2*np.sqrt(ev[1])))


# =============================================================================
# INDEPENDENT WIDTH CHECKS / SPATIAL POWER APERTURES
# =============================================================================
def ellipse_weights(shape, p, scale):
    S = shape_matrix(p)
    ex, ey = scale*np.sqrt(S[0, 0]), scale*np.sqrt(S[1, 1])
    h, w = shape
    x1, x2 = max(0, int(np.floor(p[1]-ex-1))), min(w, int(np.ceil(p[1]+ex+2)))
    y1, y2 = max(0, int(np.floor(p[2]-ey-1))), min(h, int(np.ceil(p[2]+ey+2)))
    yy, xx = np.mgrid[y1:y2, x1:x2]
    Q = np.linalg.inv(S)
    weights = np.zeros(xx.shape, dtype=float)
    offsets = (np.arange(APERTURE_SUBPIXELS)+0.5)/APERTURE_SUBPIXELS - 0.5
    for dx in offsets:
        for dy in offsets:
            a, b = xx+dx-p[1], yy+dy-p[2]
            weights += (Q[0, 0]*a*a + 2*Q[0, 1]*a*b + Q[1, 1]*b*b <= scale**2)
    full = np.zeros(shape, dtype=float)
    full[y1:y2, x1:x2] = weights / APERTURE_SUBPIXELS**2
    clipped = p[1]-ex < -0.5 or p[1]+ex > w-0.5 or p[2]-ey < -0.5 or p[2]+ey > h-0.5
    return full, bool(clipped)


def direct_width(axis, profile, center, radius):
    # Nearest crossings around the central peak; NO smoothing or tail normalization.
    candidates = np.where(np.abs(axis-center) <= max(2.0, 0.5*radius))[0]
    if not candidates.size:
        return np.nan, np.nan, np.nan
    i = int(candidates[np.argmax(profile[candidates])])
    level = profile[i]*np.exp(-2)
    if level <= 0:
        return np.nan, np.nan, np.nan
    left = right = np.nan
    for j in range(i-1, -1, -1):
        if profile[j] <= level < profile[j+1]:
            left = axis[j]+(level-profile[j])*(axis[j+1]-axis[j])/(profile[j+1]-profile[j]); break
    for j in range(i, len(axis)-1):
        if profile[j] > level >= profile[j+1]:
            right = axis[j]+(level-profile[j])*(axis[j+1]-axis[j])/(profile[j+1]-profile[j]); break
    return float(right-left), float(left), float(right)


def profile_checks(image, fit):
    p = fit["p"]
    x1, y1, x2, y2 = fit["bounds"]
    x, y = np.arange(x1, x2, dtype=float), np.arange(y1, y2, dtype=float)
    px = map_coordinates(image, [np.full_like(x, p[2]), x], order=1, mode="nearest")
    py = map_coordinates(image, [y, np.full_like(y, p[1])], order=1, mode="nearest")
    px -= background_local(p, x, p[2], fit["anchor"], fit["half"])
    py -= background_local(p, p[1], y, fit["anchor"], fit["half"])
    Q = np.linalg.inv(shape_matrix(p))
    wx = direct_width(x, px, p[1], 1/np.sqrt(Q[0, 0]))[0] * pixel_scale()[0]
    wy = direct_width(y, py, p[2], 1/np.sqrt(Q[1, 1]))[0] * pixel_scale()[1]
    records = []
    for axis_name, axis, observed, cx, sc, model in [
        ("X", x, px, p[1], pixel_scale()[0], gaussian_pixel(p, x, np.full_like(x, p[2]))),
        ("Y", y, py, p[2], pixel_scale()[1], gaussian_pixel(p, np.full_like(y, p[1]), y))]:
        records.extend(dict(axis=axis_name, position_um=float((a-cx)*sc),
                            background_corrected_counts=float(b), fitted_pixel_counts=float(c))
                       for a, b, c in zip(axis, observed, model))
    return {"direct_1e2_x_um": wx, "direct_1e2_y_um": wy}, pd.DataFrame(records)


def limited_second_moments(image, fit, good):
    # Auxiliary D4sigma in a finite 2w ellipse, with local fit plane removed.
    # Not an ISO-certified measurement; not substituted for the requested 1/e^2 widths.
    p = fit["p"]
    aperture, _ = ellipse_weights(image.shape, p, 2.0)
    yy, xx = np.indices(image.shape)
    z = (image-background_local(p, xx, yy, fit["anchor"], fit["half"])) * aperture * good
    total = z.sum()
    if total <= 0:
        return {}
    cx, cy = np.sum(z*xx)/total, np.sum(z*yy)/total
    V = np.array([[np.sum(z*(xx-cx)**2), np.sum(z*(xx-cx)*(yy-cy))],
                  [np.sum(z*(xx-cx)*(yy-cy)), np.sum(z*(yy-cy)**2)]]) / total
    sc = np.diag(pixel_scale())
    V = sc @ V @ sc
    ev = np.linalg.eigvalsh(V)
    if np.min(ev) <= 0:
        return {}
    return {"limited_D4sigma_x_um": float(4*np.sqrt(V[0, 0])),
            "limited_D4sigma_y_um": float(4*np.sqrt(V[1, 1])),
            "limited_D4sigma_minor_um": float(4*np.sqrt(ev[0])),
            "limited_D4sigma_major_um": float(4*np.sqrt(ev[1]))}


def background_sum_vector(weight):
    h, w = weight.shape
    return np.array([weight.sum(),
        np.dot(weight.sum(axis=0), (np.arange(w)-(w-1)/2)/w),
        np.dot(weight.sum(axis=1), (np.arange(h)-(h-1)/2)/h)])


def quantify_power(image, fit, total_power, noise, bg_cov):
    total_weights = rectangle_mask(image.shape, POWER_ROI_XYXY).astype(float)
    total = float(np.sum(image*total_weights))
    vtotal = background_sum_vector(total_weights)
    total_u = np.sqrt(noise**2*np.sum(total_weights**2) + vtotal @ bg_cov @ vtotal)
    records = []
    scales = sorted(set(float(s) for s in APERTURE_SWEEP_SCALES) | {float(POWER_ELLIPSE_SCALE)})
    for s in scales:
        weights, edge_cut = ellipse_weights(image.shape, fit["p"], s)
        restricted = weights * total_weights
        aperture_cut = edge_cut or np.any(weights-restricted > 1e-12)
        counts = float(np.sum(image*restricted))
        fraction_raw = counts/total if total > 0 else np.nan
        # No clipping of invalid fractions to [0,1]: flag baseline/FOV failures instead.
        valid = np.isfinite(fraction_raw) and 0 <= fraction_raw <= 1 and not aperture_cut
        fraction = fraction_raw if valid else np.nan
        delta_weights = restricted/total - counts*total_weights/total**2 if total > 0 else restricted*np.nan
        v = background_sum_vector(delta_weights)
        fraction_u = np.sqrt(noise**2*np.sum(delta_weights**2) + v @ bg_cov @ v)
        power_at_meter = fraction*total_power
        if POWER_METER_PLANE == "before_nd":
            power_before_nd = power_at_meter
        elif ND_TRANSMISSION_AT_637 is not None:
            power_before_nd = power_at_meter/ND_TRANSMISSION_AT_637
        else:
            power_before_nd = np.nan
        records.append(dict(ellipse_scale=s, aperture_counts=counts, total_counts=total,
                            raw_fraction=fraction_raw, fraction=fraction,
                            fraction_noise_indicator=float(fraction_u), clipped=bool(aperture_cut),
                            ideal_gaussian_enclosed_fraction=float(-np.expm1(-2*s*s)),
                            core_region_power_at_meter_plane_uW=power_at_meter,
                            core_region_power_before_nd_uW=power_before_nd))
    sweep = pd.DataFrame(records)
    chosen = dict(sweep.loc[np.isclose(sweep.ellipse_scale, POWER_ELLIPSE_SCALE)].iloc[0])
    p = fit["p"]
    # Infinite-domain fitted component: diagnostic Gaussian assumption only.
    gaussian_counts = float(p[0]*np.pi*np.sqrt(np.linalg.det(shape_matrix(p)))/2)
    chosen["gaussian_model_counts_infinite"] = gaussian_counts
    chosen["gaussian_model_fraction_diagnostic"] = gaussian_counts/total if total > 0 else np.nan
    chosen["total_counts_noise_indicator"] = float(total_u)
    chosen["total_counts_relative_noise_indicator"] = float(total_u/total) if total > 0 else np.inf
    power_before = chosen["core_region_power_before_nd_uW"]
    if REFERENCE_INPUT_POWER_BEFORE_ND_UW is not None and np.isfinite(power_before) and power_before > 0:
        chosen["apparent_insertion_loss_dB"] = float(-10*np.log10(power_before/REFERENCE_INPUT_POWER_BEFORE_ND_UW))
    return chosen, sweep


# =============================================================================
# DIAGNOSTICS / SAVING
# =============================================================================
def sensitivity_checks(image, good, fit, noise):
    records = []
    baseline = metrics(fit["p"])
    records.append(dict(method="primary_linear", half_size_px=fit["half"], **baseline, error=""))
    if CHECK_FIT_SENSITIVITY:
        for name, half, robust in [("smaller_roi", max(FIT_MIN_HALF_SIZE_PX, int(fit["half"]*0.8)), False),
                                    ("larger_roi", min(FIT_MAX_HALF_SIZE_PX, int(fit["half"]*1.2)), False),
                                    ("robust_fit", fit["half"], True)]:
            try:
                alternative = fit_gaussian(image, good, tuple(fit["p"][1:3]), half, noise,
                                           seed=fit["p"], robust=robust)
                records.append(dict(method=name, half_size_px=half, **metrics(alternative["p"]), error=""))
            except Exception as exc:
                records.append(dict(method=name, half_size_px=half, error=str(exc)))
    return pd.DataFrame(records)


def make_plot(raw, image, fit, summary, profiles):
    p = fit["p"]
    x1, y1, x2, y2 = fit["bounds"]
    yy, xx = np.mgrid[y1:y2, x1:x2]
    crop = image[y1:y2, x1:x2]
    model = gaussian_pixel(p, xx, yy) + background_local(p, xx, yy, fit["anchor"], fit["half"])
    fig, axes = plt.subplots(2, 3, figsize=(13, 8))
    axes[0, 0].imshow(raw, cmap="gray", origin="upper")
    axes[0, 0].plot([x1, x2, x2, x1, x1], [y1, y1, y2, y2, y1], "r-", lw=1)
    axes[0, 0].set_title("RAW frame and fit ROI")
    for ax, data, title, cmap in [(axes[0, 1], crop, "Signed background-corrected ROI", "viridis"),
                                 (axes[0, 2], model, "Pixel-integrated Gaussian + plane", "viridis"),
                                 (axes[1, 0], crop-model, "Residual: data minus fit", "RdBu_r")]:
        im = ax.imshow(data, origin="upper", extent=(x1-.5, x2-.5, y2-.5, y1-.5), cmap=cmap)
        fig.colorbar(im, ax=ax, fraction=0.046, label="counts")
        ax.set_title(title)
    S = shape_matrix(p)
    ev, U = np.linalg.eigh(S)
    t = np.linspace(0, 2*np.pi, 361)
    ellipse = U @ (np.sqrt(ev)[:, None] * np.array([np.cos(t), np.sin(t)]))
    for s, style, name in [(1., "w--", "1/e² ellipse"), (POWER_ELLIPSE_SCALE, "r-", "power aperture")]:
        axes[0, 1].plot(p[1]+s*ellipse[0], p[2]+s*ellipse[1], style, lw=1, label=name)
    axes[0, 1].legend(fontsize=8)
    for ax, axis_name in [(axes[1, 1], "X"), (axes[1, 2], "Y")]:
        d = profiles[profiles.axis == axis_name]
        ax.plot(d.position_um, d.background_corrected_counts/p[0], label="measured center cut", lw=1)
        ax.plot(d.position_um, d.fitted_pixel_counts/p[0], label="Gaussian pixel model", lw=1)
        ax.axhline(np.exp(-2), c="k", ls="--", lw=0.8, label="1/e² of fitted peak")
        ax.axvline(0, c="0.7", lw=.5)
        ax.set_xlabel("Object-plane position (µm)")
        ax.set_ylabel("Intensity / fitted continuous peak")
        ax.set_title(f"{axis_name}: MFD = {summary['mfd_'+axis_name.lower()+'_um']:.3f} µm")
        ax.legend(fontsize=7)
    for ax in axes.flat[:4]:
        ax.set_xlabel("Camera X pixel"); ax.set_ylabel("Camera Y pixel")
    fig.suptitle(f"{summary['file_name']} | {WAVELENGTH_NM:g} nm | "
                 f"MFD: {summary['mfd_quality']} | power: {summary['power_quality']}", fontsize=11)
    fig.tight_layout()
    return fig


def json_safe(value):
    if isinstance(value, dict):
        return {str(k): json_safe(v) for k, v in value.items()}
    if isinstance(value, (list, tuple, np.ndarray)):
        return [json_safe(v) for v in value]
    if isinstance(value, (np.integer,)):
        return int(value)
    if isinstance(value, (np.bool_,)):
        return bool(value)
    if isinstance(value, (float, np.floating)):
        return float(value) if np.isfinite(value) else None
    if isinstance(value, Path):
        return str(value)
    return value


def analyze_raw(raw_path, total_power, dark=None, flat=None, bad=None):
    raw, raw_info = load_raw_info(raw_path)
    for name, arr in [("dark", dark), ("flat field", flat), ("bad mask", bad)]:
        if arr is not None and arr.shape != raw.shape:
            raise ValueError(f"{name} shape {arr.shape} does not match RAW shape {raw.shape}.")
    image, good, bg_coeff, bg_cov, noise, edge = correct_frame(raw, dark, flat, bad)
    center = initial_peak(image, good)
    fit = adaptive_fit(image, good, center, noise)
    p = fit["p"]
    mm = metrics(p)
    direct, profiles = profile_checks(image, fit)
    limited = limited_second_moments(image, fit, good)
    checks = sensitivity_checks(image, good, fit, noise)
    powers, sweep = quantify_power(image, fit, total_power, noise, bg_cov)
    mfd_flags, power_flags, notes = [], [], []
    if dark is None:
        notes.append("No dark frame: electronic background and broad optical light may be inseparable.")
    if SCALE_RELATIVE_U_1SIGMA is None:
        notes.append("Scale uncertainty unknown; fit precision does not establish absolute accuracy.")
    if PSF_COVARIANCE_OBJECT_UM2 is None:
        notes.append("MFD refers to the imaged field; optical PSF/defocus is not deconvolved.")
    if not np.isfinite(total_power):
        notes.append("No meter power supplied: absolute core power and insertion loss are unavailable.")
    sat = float(2**min(RAW_VALID_BITS, raw_info["storage_bits"])-1) if SATURATION_COUNT is None else float(SATURATION_COUNT)
    saturated = (raw >= sat) & good
    power_aperture, cut = ellipse_weights(image.shape, p, POWER_ELLIPSE_SCALE)
    if np.any(saturated & (power_aperture > 0)):
        mfd_flags.append("Saturated pixels in beam region.")
    if np.any(saturated & rectangle_mask(image.shape, POWER_ROI_XYXY)):
        power_flags.append("Saturation invalidates the total/spatial power ratio.")
    if fit["at_bound"]:
        mfd_flags.append("Gaussian fit reached a parameter bound.")
    if cut:
        mfd_flags.append("Beam power ellipse intersects frame edge.")
    if fit["nrmse_core"] > .05:
        mfd_flags.append("Core residual RMS exceeds 5% of peak: non-Gaussian/structured signal.")
    if p[0]*np.exp(-2)/noise < 5:
        mfd_flags.append("1/e² intensity is less than 5 background-noise sigma.")
    if 2*np.sqrt(np.min(np.linalg.eigvalsh(shape_matrix(p)))) < 8:
        mfd_flags.append("Minor diameter is sampled by fewer than 8 camera pixels.")
    sensitivity = 0.
    for axis in ("x", "y"):
        values = checks[f"mfd_{axis}_um"].dropna().to_numpy()
        sensitivity = max(sensitivity, float(np.max(np.abs(values/mm[f"mfd_{axis}_um"]-1))))
        d = direct[f"direct_1e2_{axis}_um"]
        if not np.isfinite(d) or abs(d/mm[f"mfd_{axis}_um"]-1) > .05:
            mfd_flags.append(f"Direct {axis.upper()} 1/e² check differs by >5% or has no crossing.")
    if sensitivity > .03:
        mfd_flags.append("ROI/robust-fit width sensitivity exceeds 3%.")
    if checks.error.astype(bool).any():
        mfd_flags.append("At least one sensitivity refit failed; inspect fit_sensitivity.csv.")
    if np.any(~good):
        notes.append("Known bad pixels were excluded from fits and interpolated for power sums.")
    zero_fraction = float(np.mean(raw[edge] == 0))
    if zero_fraction > .01:
        power_flags.append("More than 1% of background-edge pixels are zero: possible black-level clipping.")
    if powers["clipped"] or not np.isfinite(powers["fraction"]):
        power_flags.append("Invalid fraction or aperture clipped by frame/meter ROI.")
    if powers["total_counts_relative_noise_indicator"] > .05:
        power_flags.append("Integrated denominator noise indicator exceeds 5%; background is inadequate.")
    if np.isfinite(powers["fraction"]) and powers["fraction_noise_indicator"] > .05*max(powers["fraction"], .01):
        power_flags.append("Spatial fraction noise indicator exceeds 5% of the fraction.")
    if POWER_METER_PLANE == "after_nd" and ND_TRANSMISSION_AT_637 is None:
        notes.append("Before-ND power and insertion loss unavailable without measured ND transmission.")
    notes.append("Spatial-aperture power can include stray/other-mode light; it is not a modal decomposition.")
    # Suppress reported power/IL if the camera ratio is demonstrably compromised.
    # Diagnostic raw fractions and aperture counts remain available for investigation.
    if power_flags:
        for k in ("core_region_power_at_meter_plane_uW", "core_region_power_before_nd_uW", "apparent_insertion_loss_dB"):
            if k in powers:
                powers[k] = np.nan
            if k in sweep:
                sweep[k] = np.nan
    sweep["power_flags"] = "; ".join(power_flags)
    summary = dict(file_name=raw_path.name, raw_path=str(raw_path.resolve()),
        status="completed", wavelength_nm=WAVELENGTH_NM, sensor_power_uW=total_power,
        raw_width_px=raw_info["width_px"], raw_height_px=raw_info["height_px"],
        raw_format=raw_info["format"], raw_storage_bits=raw_info["storage_bits"],
        camera_pixel_size_um=CAMERA_PIXEL_SIZE_UM, magnification=MAGNIFICATION,
        mfd_x_camera_px=mm["mfd_x_um"]/pixel_scale()[0],
        mfd_y_camera_px=mm["mfd_y_um"]/pixel_scale()[1],
        power_meter_plane=POWER_METER_PLANE, nd_transmission_at_637=ND_TRANSMISSION_AT_637,
        sample_um_per_pixel_x=pixel_scale()[0], sample_um_per_pixel_y=pixel_scale()[1],
        scale_relative_u1sigma=SCALE_RELATIVE_U_1SIGMA,
        center_x_px=float(p[1]), center_y_px=float(p[2]), background_noise_counts=noise,
        edge_corrected_mean_counts=float(np.mean(image[edge])),
        edge_corrected_median_counts=float(np.median(image[edge])),
        amplitude_counts=float(p[0]), core_fit_nrmse=fit["nrmse_core"],
        fit_width_sensitivity_percent=100*sensitivity, background_zero_fraction=zero_fraction,
        saturation_count=sat, saturated_pixels=int(saturated.sum()),
        dark_frames=len(DARK_RAW_PATHS), power_ellipse_scale=POWER_ELLIPSE_SCALE,
        mfd_quality="review" if mfd_flags else "checks_passed",
        power_quality="invalid_or_review" if power_flags else "conditional_on_background_and_field_match",
        mfd_flags="; ".join(mfd_flags), power_flags="; ".join(power_flags), notes="; ".join(notes),
        **mm, **fit_precision(fit), **direct, **limited, **powers, **calibrated_psf_correction(p))
    fig = make_plot(raw, image, fit, summary, profiles)
    if SAVE_OUTPUTS:
        out = raw_path.parent / raw_path.stem
        out.mkdir(exist_ok=True)
        stem = raw_path.stem
        pd.DataFrame([summary]).to_csv(out/f"{stem}_summary.csv", index=False, encoding="utf-8-sig")
        profiles.to_csv(out/f"{stem}_MFD_profiles.csv", index=False)
        sweep.to_csv(out/f"{stem}_aperture_sweep.csv", index=False)
        checks.to_csv(out/f"{stem}_fit_sensitivity.csv", index=False)
        x1, y1, x2, y2 = fit["bounds"]
        np.savez_compressed(out/f"{stem}_analysis_arrays.npz",
            signed_corrected_fit_roi=image[y1:y2, x1:x2],
            power_aperture_weight_roi=power_aperture[y1:y2, x1:x2],
            fit_parameters=p, fit_covariance=fit["covariance"], roi_xyxy=fit["bounds"],
            background_plane_coefficients=bg_coeff, background_plane_covariance=bg_cov)
        metadata = dict(raw_storage=raw_info, summary=summary, fit_parameter_order=["amplitude", "xc", "yc", "log_wprojx",
            "log_wprojy", "atanh_rho", "local_offset", "local_x_slope", "local_y_slope"],
            fit_parameters=p, fit_roi=fit["bounds"], fit_plane_anchor=fit["anchor"], fit_plane_half=fit["half"],
            dark_paths=[str(Path(d).resolve()) for d in DARK_RAW_PATHS],
            raw_sha256=hashlib.sha256(raw_path.read_bytes()).hexdigest(),
            acquisition=dict(raw_width=RAW_WIDTH, raw_height=RAW_HEIGHT, raw_dtype=RAW_DTYPE,
                byte_order=RAW_BYTE_ORDER, header_bytes=RAW_HEADER_BYTES,
                pixel_size_um=CAMERA_PIXEL_SIZE_UM, magnification=MAGNIFICATION,
                psf_covariance_object_um2=PSF_COVARIANCE_OBJECT_UM2,
                flat_field_path=FLAT_FIELD_NPY, bad_pixel_mask_path=BAD_PIXEL_MASK_NPY,
                remove_edge_background_plane=REMOVE_EDGE_BACKGROUND_PLANE,
                background_edge_fraction=BACKGROUND_EDGE_FRACTION, power_roi_xyxy=POWER_ROI_XYXY,
                reference_input_power_before_nd_uW=REFERENCE_INPUT_POWER_BEFORE_ND_UW),
            software=dict(python=platform.python_version(), numpy=np.__version__, scipy=scipy.__version__))
        (out/f"{stem}_metadata.json").write_text(json.dumps(json_safe(metadata), indent=2), encoding="utf-8")
        fig.savefig(out/f"{stem}_diagnostics.png", dpi=PNG_DPI, bbox_inches="tight")
        if SAVE_PDF:
            fig.savefig(out/f"{stem}_diagnostics.pdf", bbox_inches="tight")
    if SHOW_PLOTS:
        plt.show()
    plt.close(fig)
    print(f"{raw_path.name}: MFDx={mm['mfd_x_um']:.3f} µm, MFDy={mm['mfd_y_um']:.3f} µm | "
          f"MFD {summary['mfd_quality']} | power {summary['power_quality']}")
    for flag in mfd_flags + power_flags:
        print("  CHECK:", flag)
    return summary


# =============================================================================
# RUN ONE FILE OR EVERY RAW IN A FOLDER; WRITE THE FINAL SUMMARY
# =============================================================================
def run_analysis():
    pixel_scale()  # validate calibration
    if not isinstance(RAW_VALID_BITS, int) or not 1 <= RAW_VALID_BITS <= 16:
        raise ValueError("RAW_VALID_BITS must be the camera ADC bit depth (1 to 16).")
    if POWER_METER_PLANE not in ("before_nd", "after_nd"):
        raise ValueError("Set POWER_METER_PLANE to before_nd or after_nd.")
    if ND_TRANSMISSION_AT_637 is not None and not 0 < ND_TRANSMISSION_AT_637 <= 1:
        raise ValueError("ND transmission must be a measured fraction in (0,1].")
    if REFERENCE_INPUT_POWER_BEFORE_ND_UW is not None and REFERENCE_INPUT_POWER_BEFORE_ND_UW <= 0:
        raise ValueError("Reference input power must be positive.")
    if POWER_ELLIPSE_SCALE <= 0 or APERTURE_SUBPIXELS < 1:
        raise ValueError("Ellipse scale and subpixel count must be positive.")
    if SCALE_RELATIVE_U_1SIGMA is not None and SCALE_RELATIVE_U_1SIGMA < 0:
        raise ValueError("Scale uncertainty must be nonnegative, or None.")
    location = Path(RAW_INPUT).expanduser().resolve() if RAW_INPUT else Path.cwd()
    if location.is_file():
        if location.suffix.lower() != ".raw":
            raise ValueError("Use a linear RAW file, not JPG/display-scaled imagery.")
        files, root = [location], location.parent
    elif location.is_dir():
        excluded = {Path(p).expanduser().resolve() for p in DARK_RAW_PATHS}
        files = sorted((p for p in location.iterdir() if p.is_file() and p.suffix.lower()==".raw"
                        and p.resolve() not in excluded), key=lambda p:p.name.lower())
        root = location
    else:
        raise FileNotFoundError(f"Input file/folder does not exist: {location}")
    if not files:
        print(f"No RAW measurement files found in {root}.")
        return pd.DataFrame()
    for f in files:
        power = POWER_BY_FILENAME_UW.get(f.name, SENSOR_POWER_UW)
        if power is not None and (not np.isfinite(power) or power <= 0):
            raise ValueError(f"Enter a positive sensor power for {f.name}, or None for MFD only.")
    dark = np.mean([load_raw(Path(d).expanduser()) for d in DARK_RAW_PATHS], axis=0) if DARK_RAW_PATHS else None
    bad = np.load(BAD_PIXEL_MASK_NPY, allow_pickle=False).astype(bool) if BAD_PIXEL_MASK_NPY else None
    flat = np.load(FLAT_FIELD_NPY, allow_pickle=False).astype(float) if FLAT_FIELD_NPY else None
    for name, arr in [("dark", dark), ("bad mask", bad), ("flat field", flat)]:
        if arr is not None and arr.ndim != 2:
            raise ValueError(f"{name} must be a two-dimensional image.")
    if dark is not None and not np.isfinite(dark).all():
        raise ValueError("Dark frame contains invalid values.")
    results = []
    print(f"Processing {len(files)} RAW file(s) in {root}")
    print("MFDx/y = center cuts in camera X/Y; principal and projected widths are saved separately.")
    for f in files:
        try:
            reading = POWER_BY_FILENAME_UW.get(f.name, SENSOR_POWER_UW)
            results.append(analyze_raw(f, float(reading) if reading is not None else np.nan, dark, flat, bad))
        except Exception as exc:
            plt.close("all")
            results.append(dict(file_name=f.name, raw_path=str(f), status="failed", error=str(exc)))
            print(f"FAILED {f.name}: {exc}")
    batch = pd.DataFrame(results)
    if SAVE_OUTPUTS:
        dest = root / "batch_summary.csv"
        batch.to_csv(dest, index=False, encoding="utf-8-sig")
        print(f"Saved: {dest}")
    cols = [c for c in ["file_name", "status", "mfd_x_um", "mfd_y_um", "mfd_minor_um", "mfd_major_um",
        "direct_1e2_x_um", "direct_1e2_y_um", "fraction", "core_region_power_at_meter_plane_uW",
        "apparent_insertion_loss_dB", "mfd_quality", "power_quality", "error"] if c in batch]
    try:
        from IPython.display import display
        display(batch[cols])
    except ImportError:
        print(batch[cols].to_string(index=False))
    return batch


if __name__ == "__main__":
    batch_results = run_analysis()
```
